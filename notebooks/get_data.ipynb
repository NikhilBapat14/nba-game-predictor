```python
df["gameDateTimeEst"] = pd.to_datetime(df["gameDateTimeEst"])
```

```python
df = df.loc[df["gameDateTimeEst"] >= "1986-10-30 20:00:00"]
```

```python
hornets_teams = df[df["teamName"] == "Hornets"]
hornets_teams["teamCity"].value_counts()
```

```python
conditions = [
    (df['teamCity'] == 'Seattle') & (df['teamName'] == 'SuperSonics'),
    (df['teamCity'] == 'Washington') & (df['teamName'] == 'Bullets'),
    (df['teamCity'] == 'New Orleans') & (df['teamName'] == 'Hornets'),
    (df['teamCity'] == 'Oklahoma City') & (df['teamName'] == 'Hornets'),
    (df['teamCity'] == 'Charlotte') & (df['teamName'] == 'Bobcats'),
]

new_cities = ['Oklahoma City', 'Washington', 'New Orleans', 'New Orleans', 'Charlotte']
new_names = ['Thunder', 'Wizards', 'Pelicans', 'Pelicans', 'Hornets']

df['teamName'] = np.select(conditions, new_names, default=df['teamName'])
```

```python
df = df.loc[:, "gameId":"turnovers"]
```

```python
columns_to_drop = ["teamCity", "opponentTeamCity", "opponentTeamId", "opponentTeamName", "opponentScore"]
df = df.drop(columns=columns_to_drop)
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
missing_rows = df[df.isna().any(axis=1)]
remove_games = missing_rows["gameId"].unique()
df = df.loc[~df["gameId"].isin(remove_games)]
```

```python
df.isna().sum()
```

```python
df.teamName.value_counts()
```

```python
count_teams = df.groupby("teamName")["teamName"].transform("count")
df = df[count_teams > 3]
```
