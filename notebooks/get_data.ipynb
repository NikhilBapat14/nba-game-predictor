```python
# Calculate single game features
```

```python
def shift_opponent_stats(df):
    df = df.sort_values(by=["gameDateTimeEst", "gameId", "home"])

    opp_map = df[["gameId", "home", "reboundsDefensive"]].copy()
    
    opp_map["home"] = opp_map["home"].replace({1: 0, 0: 1})
    
    opp_map = opp_map.rename(columns={
        "reboundsDefensive": "opp_dreb"
    })
    
    df = pd.merge(df, opp_map, on=["gameId", "home"], how="left")

    return df
    
    
```

```python
def create_four_factors(df):

    df["possessions"] = .96 * (df["fieldGoalsAttempted"] + df["turnovers"] + .44 * df["freeThrowsAttempted"] - df["reboundsOffensive"])

    df["eFG"] = (df["fieldGoalsMade"] + .5 * df["threePointersMade"]) / df["fieldGoalsAttempted"]

    df["TO%"] = df["turnovers"] / df["possessions"]

    df["OREB%"] = (df["reboundsOffensive"]) / (df["reboundsOffensive"] + df["opp_dreb"])

    df["FTR"] = df["freeThrowsAttempted"] / df["fieldGoalsAttempted"]

    df = df.drop(columns=df.loc[:,"assists":"opp_dreb"].columns)

    return df
```

```python
def team_rating(df):
    df["off_rating"] = (df["points"] / df["possessions"]) * 100
    df["def_rating"] = (df["opponentScore"] / df["possessions"]) * 100
    df["net_rating"] = df["off_rating"] - df["def_rating"]
    return df
```

```python
df = shift_opponent_stats(df)
df = create_four_factors(df)
df = team_rating(df)
```
